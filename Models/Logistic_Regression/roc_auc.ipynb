```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import roc_auc_score,roc_curve
from sklearn.datasets import make_classification
import matplotlib.pyplot as plt 
```

```python
X,y = make_classification(n_samples=1000, n_classes=2,random_state=10)
```

```python
X_train, X_test, y_train, y_test = train_test_split(X,y,test_size=0.25)
```

```python
dummy_model = [0 for _ in range (len(y_test))]
dummy_model
```

```python
model = LogisticRegression()
model.fit(X_train,y_train)
```

```python
##prediction based on probability
model_prob = model.predict_proba(X_test)
model_prob
```

```python
##lets focus on positive outcome i.e taking the second column of model_prob
model_prob = model_prob[:,1]
model_prob 
```

```python
dummy_model_auc = roc_auc_score(y_test,dummy_model)
model_auc = roc_auc_score(y_test,model_prob)
print(dummy_model_auc,model_auc)
```

```python
##roc_curve will return fpr and tpr
dummy_fpr, dummy_tpr , _ = roc_curve(y_test,dummy_model)
model_fpr, model_tpr, threshold = roc_curve(y_test,model_prob)
```

```python
model_fpr
```

```python
plt.plot(dummy_fpr,dummy_tpr,linestyle='--',label='dummy_model')
plt.plot(model_fpr,model_tpr,marker='.',label='LogisticRegression')
plt.xlabel('false positive rate')
plt.ylabel('True positive rate')
plt.legend()
plt.show()
```
