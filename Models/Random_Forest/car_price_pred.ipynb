```python
from sklearn.model_selection import train_test_split
X = df.drop(columns='selling_price')
y = df['selling_price']
```

```python
##label encoding on model
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
X['model'] = le.fit_transform(X['model'])
```

```python
X.head()
```

```python
X['seller_type'].value_counts(),X['fuel_type'].value_counts(),X['transmission_type'].value_counts()
```

```python
## one hot encoding because categories very less 
num_features = X.select_dtypes(exclude='object').columns
onehot_columns = ['seller_type','fuel_type','transmission_type']

from sklearn.preprocessing import OneHotEncoder,StandardScaler
from sklearn.compose import ColumnTransformer

oh_transformer = OneHotEncoder()
scaler = StandardScaler()

preprocessor = ColumnTransformer(
    [
        ("OneHotEncoder" , oh_transformer, onehot_columns),
        ("StandardScaler", scaler, num_features)
    ],remainder='passthrough'
)
```

```python
X = preprocessor.fit_transform(X)
```

```python
pd.DataFrame(X).head()
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(X,y)
```

```python
X_train.shape,X_test.shape
```

```python
from sklearn.ensemble import RandomForestRegressor
from sklearn.linear_model import LinearRegression,Lasso,Ridge
from sklearn.tree import DecisionTreeRegressor
from sklearn.neighbors import KNeighborsRegressor
from sklearn.ensemble import AdaBoostRegressor
from sklearn.metrics import r2_score, mean_absolute_error,mean_squared_error,root_mean_squared_error
```

```python
def evaluate_model(true,predicted):
    mae = mean_absolute_error(true,predicted)
    mse = mean_squared_error(true,predicted)
    rmse = np.sqrt(mse)
    r2 = r2_score(true,predicted)
    return mae,mse,rmse,r2
```

```python
models = {
    "Random Forest" : RandomForestRegressor(),
    "LinearRegression" : LinearRegression(),
    "Lasso" : Lasso(),
    "Ridge" : Ridge(),
    "KNN" : KNeighborsRegressor(),
    "Decision Tree" : DecisionTreeRegressor(),
    "Adaboost" : AdaBoostRegressor()
}

for i in range (len(list(models))):
    model = list(models.values())[i]
    model.fit(X_train,y_train)

    y_train_pred = model.predict(X_train)
    y_test_pred = model.predict(X_test)

    model_train_mae, model_train_mse, model_train_rmse, model_train_r2 = evaluate_model(y_train,y_train_pred)
    model_test_mae, model_test_mse, model_test_rmse, model_test_r2 = evaluate_model(y_test,y_test_pred)

    print(list(models.keys())[i])

    print("Model performance for training dataset")
    print("mae" , model_train_mae)
    print("mse", model_train_mse)
    print("rmse", model_train_rmse)
    print("r2_score", model_train_r2)

    print("-------------------")

    print("Model performance for test dataset")
    print("mae" , model_test_mae)
    print("mse", model_test_mse)
    print("rmse", model_test_rmse)
    print("r2_score", model_test_r2)
    print('\n')
```
