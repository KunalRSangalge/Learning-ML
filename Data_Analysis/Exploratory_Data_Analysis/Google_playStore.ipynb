```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('googleplaystore.csv')
df.head()
```

```python
df.info()
```

1) Data Cleaning

```python
df['Reviews'].astype(int)
```

```python
#there is one value with value 3.0M in Reviews
df[~df['Reviews'].str.isnumeric()]
```

```python
df_copy = df.copy()
```

```python
df_copy = df_copy.drop(index=10472)
```

```python
df_copy[~df_copy['Reviews'].str.isnumeric()]
```

```python
df_copy['Reviews'] = df_copy['Reviews'].astype(int)
df_copy.info()
```

```python
#cleaning size
df_copy['Size'].isnull().sum()
```

```python
df_copy['Size'].unique()
```

```python
df_copy['Size'] = df_copy['Size'].str.replace('k','')
df_copy['Size'] = df_copy['Size'].str.replace('M','000')
df_copy['Size'] = df_copy['Size'].replace('Varies with device',np.nan)
```

```python
df_copy['Size'] = df_copy['Size'].astype(float)
df_copy.info()
```

```python
#installs and price
df['Installs'].unique()
```

```python
df['Price'].unique()
```

```python
char_to_remove = ['$','+',',']
col = ['Installs','Price']
for char in char_to_remove:
    for cols in col:
        df_copy[cols] = df_copy[cols].str.replace(char,'')
```

```python
df_copy['Installs'].unique()
```

```python
df_copy['Price'].unique()
```

```python
df_copy['Installs'] = df_copy['Installs'].astype(int)
df_copy['Price'] = df_copy['Price'].astype(float)
df_copy.info()
```

```python
#Last Updated
df_copy['Last Updated'].unique()
```

```python
df_copy['Last Updated'] = pd.to_datetime(df_copy['Last Updated'])
```

```python
df_copy['day'] = df_copy['Last Updated'].dt.day
df_copy['month'] = df_copy['Last Updated'].dt.month
df_copy['year'] = df_copy['Last Updated'].dt.year
```

```python
df_copy.info()
```

```python
df_copy.info()
```

```python
df_copy.to_csv('cleaned_googleplaystore.csv')
```

```python
df_copy = df_copy.drop_duplicates(subset=['App'],keep='first')
```

```python
numeric_features = [feature for feature in df_copy.columns if df_copy[feature].dtype != 'O']
categorical_features = [feature for feature in df_copy.columns if df_copy[feature].dtype == 'O']

print(numeric_features)
print(categorical_features)
```

```python
#proportion of count data on categorical columns
for col in categorical_features:
    print(df[col].value_counts(normalize=True)) #normalize -> gives fraction instead of number
    print('-------------------')
```

```python
#proportion of count data on categorical columns
plt.figure(figsize=(15,15))

for i in range(0,len(numeric_features)):
    plt.subplot(5,3,i+1)
    sns.kdeplot(x = df_copy[numeric_features[i]],shade=True,color='r')
    
```

```python
#observation: rating and year are left skewed, while review size and installs are right skewed
```

```python
plt.figure(figsize=(20,15))
category = ['Type','Content Rating']

for i in range(0,len(category)):
    plt.subplot(2,2,i+1)
    sns.countplot(x=df_copy[category[i]])
    plt.xlabel(category[i])
```
